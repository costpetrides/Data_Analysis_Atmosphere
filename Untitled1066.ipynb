## Άσκηση 1

Δίνεται το διάνυσμα:

$$
x=[3,5,8,12,7,10,15,6,9,4]
$$

**(α)** Να υπολογιστεί το άθροισμα των στοιχείων από την 3η έως την 8η θέση:

$$
S=\sum_{i=3}^{8}x_i
$$

**(β)** Να υπολογιστεί η μέση τιμή των ίδιων στοιχείων.

$$
\bar{x}=\frac{1}{6}\sum_{i=3}^{8}x_i
$$

**(γ)** Να υπολογιστεί το άθροισμα των τετραγωνικών αποκλίσεων τους από τη μέση τιμή:

$$
Q=\sum_{i=3}^{8}(x_i-\bar{x})^2
$$


```python
import numpy as np

x = np.array([3, 5, 8, 12, 7, 10, 15, 6, 9, 4])
a=x[2:8]

# (α)
S = np.sum(a)

# (β)
m = np.mean(a)

# (γ)
Q = sum((a-m)**2)

print(f"Sum = {S:.3f}")
print(f"Mean = {m:.3f}")
print(f"Squared deviations = {Q:.3f}")
```


## Άσκηση 2 — Άθροισμα γινομένων με Slicing

Δίνεται το διάνυσμα:

$$
x=[3,5,8,12,7,10,15,6,9,4]
$$

με $n=10$ παρατηρήσεις.

### (α) Επιλογή στοιχείων

Για $k=3$, να δημιουργηθούν δύο διανύσματα:

$$
a=[x_1,x_2,\ldots,x_{n-k}]
$$

$$
b=[x_{1+k},x_{2+k},\ldots,x_n]
$$

Να χρησιμοποιηθεί αποκλειστικά **Python slicing**.

### (β) Άθροισμα γινομένων

Να υπολογιστεί:

$$
\boxed{S=\sum_{i=1}^{n-k}x_i x_{i+k}}
$$

χρησιμοποιώντας τις μεταβλητές `a`, `b` και τη συνάρτηση `np.sum()`.

### (γ) Άθροισμα τετραγωνικών διαφορών

Να υπολογιστεί:

$$
\boxed{Q=\sum_{i=1}^{n-k}(x_i-x_{i+k})^2}
$$

### (δ) Μέσοι όροι

Να υπολογιστούν οι μέσοι όροι των δύο διανυσμάτων:

$$
\bar{x}_a=\frac{1}{n-k}\sum_{i=1}^{n-k}x_i
$$

$$
\bar{x}_b=\frac{1}{n-k}\sum_{i=1}^{n-k}x_{i+k}
$$

### Οδηγίες

- Χρησιμοποιήστε NumPy.
- Χρησιμοποιήστε slicing για την επιλογή στοιχείων.
- Μη χρησιμοποιήσετε βρόχους `for`.
- Εμφανίστε τα αποτελέσματα όλων των ερωτημάτων.


```python
x=np.array([3,5,8,12,7,10,15,6,9,4])

# (α)

n=len(x)
k=3

a=x[:n-k]
b=x[k:n]

print(f"a={a}")
print(f"b={b}")
```
