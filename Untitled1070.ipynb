```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Data: choose 12 or 24 months
# --------------------------------------------------
dataset = 24

y12 = np.array([
    22.2, 22.7, 32.2, 44.4, 54.8, 64.3,
    68.8, 67.1, 60.2, 49.5, 39.3, 27.4
], dtype=float)

y1987 = np.array([
    21.4, 17.9, 35.9, 47.7, 56.4, 66.3,
    70.9, 65.8, 60.1, 45.4, 39.5, 31.3
], dtype=float)

y1988 = np.array([
    20.6, 22.5, 32.9, 43.6, 56.5, 61.9,
    71.6, 69.9, 57.9, 45.2, 40.5, 26.7
], dtype=float)

if dataset == 12:
    y = y12
elif dataset == 24:
    y = np.concatenate((y1987, y1988))
else:
    raise ValueError("dataset must be 12 or 24")

# --------------------------------------------------
# Time grid and harmonic indices
# --------------------------------------------------
n = len(y)
t = np.arange(1, n + 1)          # Months 1, 2, ..., n (as in the slides)
k_values = np.arange(1, n // 2 + 1)
n_harmonics = len(k_values)
y_mean = np.mean(y)

A = np.zeros(n_harmonics)
B = np.zeros(n_harmonics)
components = np.zeros((n_harmonics, n))
y_fit = np.full(n, y_mean)

# --------------------------------------------------
# Direct harmonic analysis (A_k, B_k)
# --------------------------------------------------
for idx, k in enumerate(k_values):
    theta = 2 * np.pi * k * t / n

    # The Nyquist term for even n is counted only once.
    scale = 1 / n if (n % 2 == 0 and k == n // 2) else 2 / n

    A[idx] = scale * np.sum(y * np.cos(theta))
    B[idx] = scale * np.sum(y * np.sin(theta))
    if n % 2 == 0 and k == n // 2:
        B[idx] = 0.0

    components[idx, :] = A[idx] * np.cos(theta) + B[idx] * np.sin(theta)
    y_fit += components[idx, :]

# Amplitude, phase (in degrees), frequency and period
C = np.hypot(A, B)
phi_deg = np.mod(np.degrees(np.arctan2(B, A)), 360)
f = k_values / n                # Cycles / month
period = n / k_values          # Months / cycle

# Fraction of the total variance explained by each harmonic
R2 = np.sum(components**2, axis=1) / np.sum((y - y_mean)**2)

# --------------------------------------------------
# FFT check (adjust for t = 1,...,n rather than t = 0,...,n-1)
# --------------------------------------------------
F = np.fft.fft(y)
Z = F[k_values] * np.exp(-2j * np.pi * k_values / n)
scale = np.full(n_harmonics, 2 / n)
if n % 2 == 0:
    scale[-1] = 1 / n
A_fft = scale * Z.real
B_fft = -scale * Z.imag
if n % 2 == 0:
    B_fft[-1] = 0.0

print(f"Dataset: {n} monthly values")
print(f"Mean: {y_mean:.4f} deg F")
print(" k      A_k       B_k       C_k   phi(deg)  period    R_k^2")
for idx, k in enumerate(k_values):
    print(f"{k:2d}  {A[idx]:9.4f} {B[idx]:9.4f} {C[idx]:9.4f}"
          f"  {phi_deg[idx]:8.2f}  {period[idx]:6.2f}  {R2[idx]:8.6f}")
```
